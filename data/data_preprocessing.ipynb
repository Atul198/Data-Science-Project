## Bitcoin Preprocessing

```python
# reading data
btc = pd.read_csv('Raw/BITCOINITY/bitcoinity_data.csv')
btc.head(10)
```

```python
# shape of the dataset
btc.shape
```

```python
# information about the dataset
btc.info()
```

The data is imbalance and time parameter in object so we will convert it into timeseries with dd/mm/yyyy. Furthermore their is NaN values in our dataset that we will process and remove it.

```python
btc.describe()
```

```python
# Number of missing value available in dataset
btc.isna().sum()
```

```python
#dropping dubplicates from dataset
btc.drop_duplicates(inplace=True)
btc.head()
```

```python
#Handle Missing Values
btc_numeric = btc.select_dtypes(include='number').columns
btc[btc_numeric] = btc[btc_numeric].fillna(btc[btc_numeric].median()) 
```

```python
# the following time column have function in it so we need to change it first and then we will change the type to datetime.
type(btc['Time'].iloc[0])
```

```python
# converting data type
btc['Time'] = pd.to_datetime(
    btc['Time'],
    format='%Y-%m-%d %H:%M:%S %Z',
    utc=True,
    errors='coerce'
)
btc['Time'] = pd.to_datetime(btc["Time"].dt.strftime('%Y-%m-%d'))
btc.head()
```

```python
# Saving the bitcoin data into interim
interim_path = '../data/interim/bitcoinity_cleaned.csv'
btc.to_csv(interim_path, index=False)
```

## Crude Petroleum Oils

### Import

```python
# reading crude oil import dataset 
cpo_im = pd.read_excel(r'Raw\crude petroleum oils\Crude oil Import\TradeData.xlsx')
```

```python
# printing first 5 rows from dataset
cpo_im.head()
```

```python
#shape of the dataset
cpo_im.shape
```

```python
#information about the dataset
cpo_im.info()
```

```python
# unique values and there datatype for preprocessing
cols = cpo_im.columns

cpo_im_table = pd.DataFrame({
    'Column Name': cols,
    'Unique Value': [cpo_im[col].unique() for col in cols],
    'Count': [cpo_im[col].count() for col in cols],
    'Data Type': [cpo_im[col].dtypes for col in cols]
})

cpo_im_table
```

```python
#dropping useless columns from dataset
drop_cols_im = ['typeCode','freqCode', 'refPeriodId', 'refMonth', 'period', 
             'flowCode', 'flowDesc','partnerCode', 'partnerISO', 'partnerDesc',
             'partner2Code', 'partner2ISO', 'partner2Desc', 'classificationSearchCode',
             'isOriginalClassification', 'aggrLevel', 'isLeaf', 'customsCode', 
             'customsDesc', 'mosCode', 'motCode', 'motDesc', 'isQtyEstimated',
             'isAltQtyEstimated', 'isNetWgtEstimated', 'isGrossWgtEstimated',
             'isReported', 'isAggregate'
]

cpo_im.drop(drop_cols_im, axis=1, inplace=True)
```

```python
# NaN value in dataset
cpo_im.isna().sum()
```

```python
# replacing row's data from column
cpo_im['cmdDesc'] = cpo_im['cmdDesc'].replace('Petroleum oils and oils obtained from bituminous minerals; crude', 'Crude petroleum oil')
```

```python
# removing NaN values from dataset
cpo_im['qtyUnitAbbr'] = cpo_im['qtyUnitAbbr'].fillna(cpo_im['qtyUnitAbbr'].mode()[0])
cpo_im['altQtyUnitAbbr'] = cpo_im['altQtyUnitAbbr'].fillna(cpo_im['altQtyUnitAbbr'].mode()[0])
cpo_im['netWgt'] = cpo_im['netWgt'].fillna(cpo_im['netWgt'].median())
cpo_im['cifvalue'] = cpo_im['cifvalue'].fillna(cpo_im['cifvalue'].median())
cpo_im['fobvalue'] = cpo_im['fobvalue'].fillna(cpo_im['fobvalue'].median())
```

```python
cpo_im.info()
```

```python
cols = cpo_im.columns

cpo_im_table = pd.DataFrame({
    'Column Name': cols,
    'Unique Value': [cpo_im[col].unique() for col in cols],
    'Count': [cpo_im[col].count() for col in cols],
    'Data Type': [cpo_im[col].dtypes for col in cols]
})

cpo_im_table
```

```python
# Saving Crude petroleum oil import

interim_path = r'../data/interim/crude_pertroleum_import.csv'

cpo_im.to_csv(interim_path, index=False)
```
